```python
# Series: is a 1-dimensional array
# loc: location by caption or index
# iloc: location by integer position
data = [100, 102, 104]

series = pd.Series(data, index=["a", "b", "c"])

series.loc["c"] = 200

print(series)
print()
print(series.iloc[1])
```

```python
data = [100, 102, 104, 200,202]
series = pd.Series(data, index=['a', 'b', 'c', 'd', 'e'])

print(series[series < 200])
```

```python
calories = {
    "Day 1": 1750,
    "Day 2": 2001,
    "Day 3": 1700
}

series = pd.Series(calories)

# print(series.loc["Day 3"])
series.loc["Day 3"] += 500

print(series[series >= 2000])
print()

print(series)
```

```python
# DATAFRAMES: a tabular data structure with rows and columsn

data = {
    "Name": ["Spongebob", "Patrick", "Sandy"],
    "age": [30, 35, 50]
}

dataframe = pd.DataFrame(data, index = ["Employee 1", "Employee 2", "Employee 3"])

# print(dataframe.loc["Employee 2"])
print(dataframe.iloc[0])
```

```python
# Add a new column
dataframe['Job'] = ['cook', 'N/A', 'scientist']
# print(dataframe)

# Add a new row: create a new dataframe and combine the old and the new
new_row = pd.DataFrame([{"Name": "Squidward", "age": 34, "Job": "cashier"},
                        {"Name": "Eugene", "age": 60, "Job": "manager"}], 
                       index=["Employee 4", "Employee 5"])
dataframe = pd.concat([dataframe, new_row])

print(dataframe)
```
